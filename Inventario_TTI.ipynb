## Creacion de calculos

```python
df_invetariotti["Costo con landeo Real"] = df_invetariotti["Total Landeo"] * df_invetariotti["2026 Prelim Transfer Price"]* 1500
```

```python
df_invetariotti["Costo Real"] = df_invetariotti["Costo con landeo Real"] * df_invetariotti["Ventas del mes"]
```

```python
df_invetariotti.columns
```

```python
df_invetariotti["Costo"] = df_invetariotti["Costo Standard  Unitario en $"] * df_invetariotti["Ventas del mes"]
```

```python
df_invetariotti["Rentabilidad absoluta"] = df_invetariotti["VENTA"] - df_invetariotti["Costo Real"]
```

```python
df_invetariotti["Rentabilidad %"] = (df_invetariotti["VENTA"] - df_invetariotti["Costo"])/df_invetariotti["VENTA"]
```
